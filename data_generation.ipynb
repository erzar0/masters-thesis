```python
from torch import nn
import torch.nn.functional as F
import torch



target_tensor = torch.tensor(X[1001], dtype=torch.float32)
input_tensor = torch.tensor( X[10], dtype=torch.float32)

plt.plot(target_tensor)
plt.plot(input_tensor)

# Normalize target to be a probability distribution
target_tensor /= target_tensor.sum()

# Convert input to log probabilities using log_softmax
input_log_prob = F.log_softmax(input_tensor, dim=0)

# Compute KL Divergence loss
kl_loss = F.kl_div(input_log_prob, target_tensor, reduction='batchmean')  # 'batchmean' ensures proper averaging

print("KL Divergence Loss:", kl_loss.item())
```
